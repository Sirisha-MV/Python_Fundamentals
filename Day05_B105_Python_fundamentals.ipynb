```python
organizing list datatype:
```

```python
# i want to organize the list in the alphabetical order ---> A-Z
```

```python
1. temporary approach ---> sorted ---> we will be having the original order of list
2.permanent approach ---> sort ---> changes are implied permanently
```

```python
print(sorted(cars)) # A-Z
```

```python
print(cars) # this is the original order
```

```python
cars.sort()
```

```python
print(cars)
```

```python
**Interview question: what is the difference between sorted and sort method in the list data type?
sorted in temporary operation applied on the list where are sort is permenent action on the list which cannot be reversed to original order of the list
```

```python
useraccount = {'username':'sirisha mv', 'firstname':'sirisha', 'lastname':'mv','DOB':'22-12-1999', 'pwd': '99660' }
```

```python
for a,b in useraccount.items():
    print(a)
    print(b)
```

```python
for a,b in useraccount.items():
    print(f"key:{a}")
    print(f"value:{b}")
```

```python
for a,b in useraccount.items():
    print(f"key:{a}")
    print(f"value:{b}\n")
```

```python
#req: i want only keys..?
```

```python
for a in useraccount.keys():
    print(a)
```

```python
for a in useraccount.keys():
    print(f"key: {a}")
```

```python
#req: i want only values..?
```

```python
for b in useraccount.values():
    print(b)
```

```python
for b in useraccount.values():
    print(f"value: {b}")
```
